```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 100)
y = np.sin(x) + 0.5*np.sin(3*x) + 0.3*np.random.randn(len(x))
plt.plot(x, y)
plt.title("Simple Sine Wave")
plt.show()
```
