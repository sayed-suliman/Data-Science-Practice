```python
import pandas as pd

data = {
    "product": ["apples", "bananas", "oranges"],
    "profit": [20, 15, 25]
}

df = pd.DataFrame(data, index=["a", "b", "c"])
print(df)
print('-----------------------------')
getdata = df.loc['b']
getdata = df.loc[['c','b'],'product']
getdata = df.loc[df['profit'] > 15]
df.loc['c','profit'] = 50
getdata = df.loc['c']
print(getdata)
```

```python
import matplotlib.pyplot as plt

x = [1, 10, 100, 1000]
y = [10, 20, 30, 40]

plt.scatter(x, y)

plt.xscale('log')   # 👈 important
plt.xlabel("X values (log scale)")
plt.ylabel("Y values")
plt.title("Example Log Scale")

plt.show()
```
